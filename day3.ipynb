```python
# Class keyword

class Cat:
    attribute1 = "Animal"
    attribute2 = "Hairy"

    def fun(self):
        print('It is an', self.attribute1)
        print('It is', self.attribute2)
White = Cat()
print(White.attribute1)
print(White.attribute2)
White.fun()
```
